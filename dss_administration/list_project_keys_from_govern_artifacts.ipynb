```python
from dataikuapi.govern.artifact_search import GovernArtifactSearchQuery

# Define connection parameters
host = "https://YOUR-GOVERN-HOST"
api_key = "REDACTED"

# Initialize the client
client = GovernClient(host, api_key)
```

```python
# Setup a basic search query for all artifacts
artifacts_query = GovernArtifactSearchQuery()

# Create a search request
search_request = client.new_artifact_search_request(artifacts_query)

project_keys = set()

# Perform the search with potential batched results
try:
    while True:
        # Fetch next batch
        result_batch = search_request.fetch_next_batch()
        artifacts_list = result_batch.get_raw().get('uiArtifacts',[])
        # print(artifacts_list[0].get('artifact',{}))
        if not artifacts_list:
            # If there are no more results, break
            break

        # Process each artifact in the result batch
        for artifact_hit in artifacts_list:
            artifact_hit = artifact_hit.get('artifact',{})
            pk = artifact_hit.get('fields',{}).get('project_key',None)
            if pk:
                project_keys.add(pk)
            #print(f"Artifact ID: {artifact_hit['id']}, Name: {artifact_hit['name']} Project_key: {}")
except Exception as e:
    print(f"Error during searching or fetching artifacts: {e}")

```
